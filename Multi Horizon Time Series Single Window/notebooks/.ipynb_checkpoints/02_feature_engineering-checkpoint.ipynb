## Imports and Config

```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt

# ── Configuration 
DATA_PATH      = '../data/raw/'
PROCESSED_PATH = '../data/processed/'
N_WEEKS        = 17    # Use first 17 weeks

FEATURE_NAMES = [
    'F1_login_count',
    'F2_vle_clicks',
    'F3_avg_score',
    'F4_num_submitted',
    'F5_on_time_rate',
    'F6_inactive_days',
    'F7_score_change',
    'F8_running_avg',
    'F9_score_variance',
    'F10_min_score',
    'F11_max_score',
    'F12_late_count',
    'F13_weeks_since_active',
]

print(f"Config ready.")
print(f"Using {N_WEEKS} weeks and "
      f"{len(FEATURE_NAMES)} features.")
```

## Load Data

```python
student_info = pd.read_csv(DATA_PATH + 'studentInfo.csv')
student_ass  = pd.read_csv(DATA_PATH + 'studentAssessment.csv')
assessments  = pd.read_csv(DATA_PATH + 'assessments.csv')
student_vle  = pd.read_csv(DATA_PATH + 'studentVle.csv')

print(f"Students:    {len(student_info):,}")
print(f"Assessments: {len(student_ass):,}")
print(f"VLE records: {len(student_vle):,}")
```

## Create Labels

```python
student_info['label'] = student_info[
    'final_result'].apply(
    lambda x: 0 if x in ['Pass','Distinction'] else 1)

print(student_info['label'].value_counts())
print(f"At-risk rate: {student_info['label'].mean()*100:.1f}%")
```

## Prepare Assessment Data

```python
# Merge scores with due dates
ass = student_ass.merge(
    assessments[['id_assessment','date',
                 'assessment_type','weight']],
    on='id_assessment', how='left')

# Remove exam rows
ass = ass.dropna(subset=['date'])

# Convert days to weeks
ass['week'] = (ass['date'] / 7).astype(int) + 1

# Fill missing scores with 0
ass['score'] = ass['score'].fillna(0)

# On time flag
ass['on_time'] = (
    ass['date_submitted'] <= ass['date']
).astype(int)

# Late flag
ass['is_late'] = (
    ass['date_submitted'] > ass['date']
).astype(int)

# Keep only weeks within window
ass = ass[ass['week'] <= N_WEEKS]

print(f"Assessment records within "
      f"{N_WEEKS} weeks: {len(ass):,}")
```

## Prepare VLE Data

```python
vle = student_vle.copy()
vle['week'] = (vle['date'] / 7).astype(int) + 1
vle = vle[vle['week'] <= N_WEEKS]

print(f"VLE records within "
      f"{N_WEEKS} weeks: {len(vle):,}")
```

## Group VLE (F1 and F2)

```python
vle_weekly = vle.groupby(
    ['id_student','code_module',
     'code_presentation','week'],
    as_index=False
).agg(
    F1_login_count = ('sum_click', 'count'),
    F2_vle_clicks  = ('sum_click', 'sum')
)

print(f"VLE weekly groups: {len(vle_weekly):,}")
print(vle_weekly.head(3))
```

## Group Assessments (F3 to F12)

```python
ass_weekly = ass.groupby(
    ['id_student','week'],
    as_index=False
).agg(
    F3_avg_score       = ('score',          'mean'),
    F4_num_submitted   = ('id_assessment',  'count'),
    F5_on_time_rate    = ('on_time',        'mean'),
    days_with_activity = ('date_submitted', 'nunique'),
    F9_score_variance  = ('score',          'std'),
    F10_min_score      = ('score',          'min'),
    F11_max_score      = ('score',          'max'),
    F12_late_count     = ('is_late',        'sum')
)

# F6 inactive days
ass_weekly['F6_inactive_days'] = (
    7 - ass_weekly['days_with_activity']
).clip(lower=0)

# Fill variance with 0 when only one submission
ass_weekly['F9_score_variance'] = \
    ass_weekly['F9_score_variance'].fillna(0)

# Remove helper column
ass_weekly = ass_weekly.drop(
    columns=['days_with_activity'])

print(f"Assessment weekly groups: {len(ass_weekly):,}")
print(ass_weekly.head(3))
```

## Build Student-Week Grid

```python
# Valid students with at least 2 weeks of data
valid_ids = ass_weekly[
    ass_weekly['F4_num_submitted'] > 0
].groupby('id_student')['week'].nunique()
valid_ids = valid_ids[valid_ids >= 2].index

students = student_info[
    student_info['id_student'].isin(valid_ids)
][['id_student','code_module',
   'code_presentation','label']].drop_duplicates()

print(f"Valid students: {len(students):,}")

# Every student x every week
all_weeks = pd.DataFrame(
    {'week': range(1, N_WEEKS + 1)})
grid = students.merge(all_weeks, how='cross')

print(f"Grid size: {len(grid):,} rows")
```

## Merge Features Into Grid

```python
# Add assessment features
grid = grid.merge(
    ass_weekly[[
        'id_student','week',
        'F3_avg_score','F4_num_submitted',
        'F5_on_time_rate','F6_inactive_days',
        'F9_score_variance','F10_min_score',
        'F11_max_score','F12_late_count']],
    on=['id_student','week'], how='left')

# Add VLE features
grid = grid.merge(
    vle_weekly[[
        'id_student','code_module',
        'code_presentation','week',
        'F1_login_count','F2_vle_clicks']],
    on=['id_student','code_module',
        'code_presentation','week'],
    how='left')

# Sort by student then week
grid = grid.sort_values(
    ['id_student','week']
).reset_index(drop=True)

print(f"Grid after merge: {len(grid):,} rows")
```

## Fill Missing Values

```python
# No VLE activity = 0
grid['F1_login_count'] = grid['F1_login_count'].fillna(0)
grid['F2_vle_clicks']  = grid['F2_vle_clicks'].fillna(0)

# Score — carry forward, default 50
grid['F3_avg_score'] = grid.groupby(
    'id_student')['F3_avg_score'].transform(
    lambda x: x.ffill().fillna(50.0))

# No submissions = 0
grid['F4_num_submitted']  = grid['F4_num_submitted'].fillna(0)

# No assessment = on time by default
grid['F5_on_time_rate']   = grid['F5_on_time_rate'].fillna(1.0)

# No activity = full week inactive
grid['F6_inactive_days']  = grid['F6_inactive_days'].fillna(7.0)

# Score variance = 0 if no submissions
grid['F9_score_variance'] = grid['F9_score_variance'].fillna(0)

# Min/max carry forward
grid['F10_min_score'] = grid.groupby(
    'id_student')['F10_min_score'].transform(
    lambda x: x.ffill().fillna(50.0))
grid['F11_max_score'] = grid.groupby(
    'id_student')['F11_max_score'].transform(
    lambda x: x.ffill().fillna(50.0))

# Late count = 0 if no submissions
grid['F12_late_count'] = grid['F12_late_count'].fillna(0)

print("Missing values filled.")
```

## Calculate F7, F8, F13

```python
# F7 — score change from last week
grid['F7_score_change'] = grid.groupby(
    'id_student')['F3_avg_score'].diff().fillna(0)

# F8 — running average score
grid['F8_running_avg'] = grid.groupby(
    'id_student')['F3_avg_score'].transform(
    lambda x: x.expanding().mean())

# F13 — weeks since last submission
grid['last_active_week'] = grid.groupby(
    'id_student').apply(
    lambda g: g['week'].where(
        g['F4_num_submitted'] > 0
    ).ffill()
).reset_index(level=0, drop=True)

grid['F13_weeks_since_active'] = (
    grid['week'] - grid['last_active_week']
).fillna(0).clip(lower=0)

grid = grid.drop(columns=['last_active_week'])

print("F7, F8, F13 calculated.")

# Show example student
example = grid[
    grid['id_student'] == grid['id_student'].iloc[0]
][['week','F3_avg_score',
   'F7_score_change',
   'F8_running_avg',
   'F13_weeks_since_active']].head(10)
print("\nExample student:")
print(example.to_string(index=False))
```

## Build X and Y Arrays

```python
# Keep students with exactly N_WEEKS rows
week_counts   = grid.groupby('id_student')['week'].count()
full_students = week_counts[week_counts == N_WEEKS].index
grid          = grid[grid['id_student'].isin(full_students)]

n_students = len(full_students)
n_features = len(FEATURE_NAMES)

print(f"Students with {N_WEEKS} complete weeks: "
      f"{n_students:,}")

# Build X — shape (students, weeks, features)
X = grid[FEATURE_NAMES].values.reshape(
    n_students, N_WEEKS, n_features
).astype(np.float32)

# Build y — shape (students,)
y = grid.groupby('id_student')['label'].first()
y = y.loc[full_students].values.astype(np.float32)

print(f"\nX shape: {X.shape}")
print(f"y shape: {y.shape}")
print(f"At-risk: {y.mean()*100:.1f}%")
```

## Clean Bad Values

```python
print(f"NaN before: {np.isnan(X).sum()}")
print(f"Inf before: {np.isinf(X).sum()}")

X = np.nan_to_num(X,
                   nan=0.0,
                   posinf=100.0,
                   neginf=0.0)

print(f"NaN after:  {np.isnan(X).sum()}")
print(f"Inf after:  {np.isinf(X).sum()}")
print("Data is clean.")
```

##  Save Files

```python
os.makedirs(PROCESSED_PATH, exist_ok=True)

np.save(PROCESSED_PATH + 'X_sequences.npy', X)
np.save(PROCESSED_PATH + 'y_labels.npy', y)

print("=" * 50)
print("SAVED SUCCESSFULLY")
print("=" * 50)
print(f"X_sequences.npy → {X.shape}")
print(f"y_labels.npy    → {y.shape}")
print()
print("Features:")
for i, name in enumerate(FEATURE_NAMES):
    tag = "NEW" if i >= 8 else "   "
    print(f"  {tag} {i+1:>2}. {name}")
print()
print("Now update data_loader.py:")
print("  Change N_WEEKS = 17")
print("Then run all model files again.")
```
